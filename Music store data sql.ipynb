```python
#Load all datasets
import os
import pandas as pd
from sqlalchemy import create_engine

# MySQL connection
engine = create_engine("mysql+pymysql://root:12345@localhost/my")

folder_path = r"E:\Inntrvu_projects_data\SQL Assignment\SQL_Music_Store_Analysis-main\music store data"

for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        file_path = os.path.join(folder_path, file)

        # Read CSV
        df = pd.read_csv(file_path)

        # Create table name from file name (remove .csv)
        table_name = os.path.splitext(file)[0].lower().replace(" ", "_")

        # Load into MySQL
        df.to_sql(
            name=table_name,
            con=engine,
            if_exists="replace",  # replace if table exists
            index=False
        )

        print(f"{file} loaded into table '{table_name}' successfully")
```
